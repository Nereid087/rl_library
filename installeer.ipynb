# Installeer de Rijnland-library

Dit notebook installeert `rijnland_core` in je venv. Daarna kun je de library in elk notebook in deze repository gebruiken, bijvoorbeeld:

```python
from rijnland_core.arcgis_downloader import haal_laag_op
```

1. Klik rechtsboven in dit notebook op **Select Kernel**. Kies **Python Environments** en daarna dezelfde venv als in stap 4 van de README.
2. Klik in de eerste cel hieronder en druk op `Shift+Enter`. Controleer: er staat `Goed: je zit in een venv.` en het pad wijst naar je venv uit stap 4. Staat er `Let op`? Ga dan terug naar 1.
3. Klik in de tweede cel en druk op `Shift+Enter`.
4. Wacht tot de cel klaar is. Controleer: onderaan staat `Successfully installed rl-library-0.1.0`.

Je hoeft dit maar één keer te doen. Had je al een ander notebook open? Klik daar dan bovenin op **Restart**, anders vindt dat notebook de library nog niet.

````python
import sys

print(sys.executable)                   # de Python waar dit notebook in draait

if sys.prefix != sys.base_prefix:       # in een venv verschillen deze twee
    print("Goed: je zit in een venv.")
else:
    print("Let op: dit is géén venv. Kies via Select Kernel de venv uit stap 4.")
````
